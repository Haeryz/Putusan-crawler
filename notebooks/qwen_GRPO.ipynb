### Unsloth

Goal: GRPO-tune `DeepSeek-R1-0528-Qwen3-8B` into a **windowed line-anchored putusan extractor** (Stage 2, Plan B) using the `grpo` config of [`Haeryz/putusan-windowed-extraction`](https://huggingface.co/datasets/Haeryz/putusan-windowed-extraction) — the same repo the SFT notebook trains from (`sft` config), same document-disjoint splits as the legacy dataset. The model reads ONE line-numbered window (`NNNNN|content`) and must emit `{"sections": {key: [[start, end], ...]}, "sections_absent": [...]}` using the GLOBAL line numbers printed in the window.

Every reward is **verifiable in pure Python**: JSON validity + canonical-key structure, in-window range validity, per-section line-overlap F1 against the gold ranges, and a strict exact-match bonus. (The old `langid` language reward is gone — the output is a line-range JSON, not prose.) A deterministic assembler (`notebooks/build_windowed_dataset.py::assemble_document`) later merges window predictions back into the 31-section verbatim JSON — the model never generates court text, so post-assembly hallucination is structurally impossible.

````python
from unsloth import FastLanguageModel
import torch
# Plan B: window prompts are <= ~8K tokens and completions are a short line-range JSON
# (plus the R1 <think> trace), so 16K total context is plenty — half the old 32K,
# which frees VRAM for more rollouts per step on an A100 40GB.
max_seq_length = 16384
lora_rank = 32 # Larger rank = smarter, but slower

model, tokenizer = FastLanguageModel.from_pretrained(
    model_name = "unsloth/DeepSeek-R1-0528-Qwen3-8B",
    max_seq_length = max_seq_length,
    load_in_4bit = True, # False for LoRA 16bit
    fast_inference = True, # Enable vllm fast inference
    max_lora_rank = lora_rank,
    gpu_memory_utilization = 0.9, # Reduce if out of memory
)

model = FastLanguageModel.get_peft_model(
    model,
    r = lora_rank, # Choose any number > 0 ! Suggested 8, 16, 32, 64, 128
    target_modules = [
        "q_proj", "k_proj", "v_proj", "o_proj",
        "gate_proj", "up_proj", "down_proj",
    ],
    lora_alpha = lora_rank*2, # *2 speeds up training
    use_gradient_checkpointing = "unsloth", # Reduces memory usage
    random_state = 3407,
)
````

We create a regex format to match the reasoning sections and answers:

````python
import re

# Add optional EOS token matching
solution_end_regex = rf"{reasoning_end}(.*)"

match_format = re.compile(solution_end_regex, re.DOTALL)
match_format
````

We verify it works:

````python
match_format.findall(
    "Let me think!</think>"\
    f"Hence, the solution is 2.",
)
````

````python
match_format.findall(
    "<think>Let me think!</think>"\
    f"\n\nHence, the solution is 2",
)
````

We now want to create a reward function to match the format exactly - we reward it with 3 points if it succeeds:

````python
def match_format_exactly(completions, **kwargs):
    scores = []
    for completion in completions:
        score = 0
        response = completion[0]["content"]
        # Match if format is seen exactly!
        if match_format.search(response) is not None: score += 3.0
        scores.append(score)
    return scores
````

If it fails, we want to reward the model if it at least follows the format partially, by counting each symbol:

````python
def match_format_approximately(completions, **kwargs):
    scores = []
    for completion in completions:
        score = 0
        response = completion[0]["content"]
        # Count how many keywords are seen - we penalize if too many!
        # If we see 1, then plus some points!

        # No need to reward the think tag since we always prepend it!
        score += 0.5 if response.count(reasoning_start) == 1 else -1.0
        score += 0.5 if response.count(reasoning_end)   == 1 else -1.0
        scores.append(score)
    return scores
````

We extract the generated answer (the text after the reasoning block) and reward **structure**: is it parseable JSON with a `sections` dict of line ranges, do `sections` + `sections_absent` together cover the 31 canonical keys, and are there no invented keys?

````python
def _extract_json_text(response):
    """Take the text after the reasoning block (or the whole response) and strip code fences."""
    m = match_format.search(response)
    text = (m.group(1) if m is not None else response).strip()
    if text.startswith("```"):
        text = re.sub(r"^```(?:json)?\s*", "", text)
        text = re.sub(r"\s*```$", "", text)
    return text

def _parse_target(text):
    """Parse a model response into {"sections": dict, ...}, or None if malformed."""
    try:
        obj = json.loads(text)
    except Exception:
        return None
    if not isinstance(obj, dict) or not isinstance(obj.get("sections"), dict):
        return None
    return obj

def _as_ranges(value):
    """Normalize a predicted section value into [(start, end), ...] int pairs."""
    out = []
    if not isinstance(value, list):
        return out
    for r in value:
        if isinstance(r, (list, tuple)) and len(r) == 2:
            try:
                s, e = int(r[0]), int(r[1])
            except (TypeError, ValueError):
                continue
            if 1 <= s <= e:
                out.append((s, e))
    return out

def check_answer(prompts, completions, answer, **kwargs):
    """Structure reward: valid line-range JSON (+2), canonical coverage of
    sections+sections_absent (up to +2), invented keys / present-and-absent overlap penalized."""
    canonical = set(CANONICAL_SECTIONS)
    scores = []
    for completion in completions:
        response = completion[0]["content"]
        pred = _parse_target(_extract_json_text(response))
        if pred is None:
            scores.append(-4.0)
            continue
        score = 2.0
        present = set(pred["sections"].keys())
        absent = set(pred.get("sections_absent") or []) if isinstance(pred.get("sections_absent"), list) else set()
        covered = (present | absent) & canonical
        score += 2.0 * len(covered) / len(canonical)
        extras = len((present | absent) - canonical)
        score -= 0.5 * min(extras, 4)
        score -= 0.5 * min(len(present & absent), 2)  # a key can't be both
        scores.append(score)
    return scores
````

Structure alone isn't enough — the **content** reward compares predicted line ranges against the gold ranges per section (mean line-overlap F1 over the 31 canonical sections), plus a bonus for staying **inside the window**: every predicted range must lie within the window's printed `line_start`..`line_end` (the line numbers the model can actually see — anything else is fabricated).

````python
def _line_set(ranges, lo, hi):
    """Expand ranges (clipped to [lo, hi]) into a set of line numbers."""
    out = set()
    for s, e in ranges:
        s, e = max(s, lo), min(e, hi)
        if s <= e:
            out.update(range(s, e + 1))
    return out

def _section_line_f1(pred_sections, gold_sections_dict, lo, hi):
    """Mean per-section F1 of line-number overlap over the 31 canonical sections."""
    f1s = []
    for key in CANONICAL_SECTIONS:
        g = _line_set(_as_ranges(gold_sections_dict.get(key, [])), lo, hi)
        p = _line_set(_as_ranges(pred_sections.get(key, [])), lo, hi)
        if not g and not p:
            f1s.append(1.0)
        elif not g or not p:
            f1s.append(0.0)
        else:
            tp = len(g & p)
            precision, recall = tp / len(p), tp / len(g)
            f1s.append(0.0 if tp == 0 else 2 * precision * recall / (precision + recall))
    return sum(f1s) / len(f1s)

def check_sections(prompts, completions, answer, line_start, line_end, **kwargs):
    """Content reward: mean line-overlap F1 vs gold (up to +5) + in-window validity bonus (-1 .. +2)."""
    scores = []
    for completion, gold_text, lo, hi in zip(completions, answer, line_start, line_end):
        response = completion[0]["content"]
        pred = _parse_target(_extract_json_text(response))
        if pred is None:
            scores.append(-2.0)
            continue
        gold = gold_target(gold_text)["sections"]
        score = 5.0 * _section_line_f1(pred["sections"], gold, lo, hi)
        all_ranges = [r for key in CANONICAL_SECTIONS for r in _as_ranges(pred["sections"].get(key, []))]
        if all_ranges:
            in_window = sum(lo <= s and e <= hi for s, e in all_ranges) / len(all_ranges)
            score += 2.0 * in_window - 1.0
        scores.append(score)
    return scores
````

We now prepare our main function which periodically prints a generated response next to the gold answer, and gives a strict bonus when the predicted line ranges match the gold target exactly (same ranges per section, same absent set).

````python
global PRINTED_TIMES
PRINTED_TIMES = 0
global PRINT_EVERY_STEPS
PRINT_EVERY_STEPS = 5

def _normalized_sections(target):
    """{key: sorted list of (start, end)} for every canonical key (missing -> [])."""
    secs = target.get("sections") or {}
    return {key: sorted(_as_ranges(secs.get(key, []))) for key in CANONICAL_SECTIONS}

def check_exact_sections(prompts, completions, answer, **kwargs):
    """Strict reward: +3 only when every canonical section's line ranges match gold exactly."""
    question = prompts[0][-1]["content"]
    responses = [completion[0]["content"] for completion in completions]

    # Print only every few steps
    global PRINTED_TIMES
    global PRINT_EVERY_STEPS
    if PRINTED_TIMES % PRINT_EVERY_STEPS == 0:
        print(
            '*'*20 + f"Question (start):\n{question[:600]}",
            f"\nGold answer (start):\n{answer[0][:600]}",
            f"\nResponse (start):\n{responses[0][:600]}",
        )
    PRINTED_TIMES += 1

    scores = []
    for response, gold_text in zip(responses, answer):
        pred = _parse_target(_extract_json_text(response))
        if pred is None:
            scores.append(-1.0)
            continue
        gold = gold_target(gold_text)
        exact = _normalized_sections(pred) == _normalized_sections(gold)
        scores.append(3.0 if exact else 0.0)
    return scores
````

Get the top 90% prompt length so we don't accidentally truncate them!

Ie we'll remove the top 10% long prompts.

````python
tokenized = dataset.map(
    lambda x: {"tokens" : tokenizer.apply_chat_template(x["prompt"], add_generation_prompt = True, tokenize = True)},
    batched = True,
)
tokenized = tokenized.map(lambda x: {"L" : len(x["tokens"])})

import numpy as np
maximum_length = int(np.quantile(tokenized["L"], 0.9))
print("p90 prompt length =", maximum_length)

# Windowed prompts are built to fit (~8K tokens max), so this filter should keep
# essentially every row — it remains as a safety net only.
maximum_length = min(maximum_length, max_seq_length // 2)
print("Prompt token budget =", maximum_length)

# Keep only samples that fit the prompt budget
dataset = dataset.select(np.where(np.array(tokenized["L"]) <= maximum_length)[0])
print("Rows kept after length filter:", len(dataset))
del tokenized
````

````python
max_prompt_length = maximum_length + 1 # + 1 just in case!
max_completion_length = max_seq_length - max_prompt_length

from vllm import SamplingParams
vllm_sampling_params = SamplingParams(
    min_p = 0.1,
    top_p = 1.0,
    top_k = -1,
    seed = 3407,
    stop = [tokenizer.eos_token],
    include_stop_str_in_output = True,
)

from trl import GRPOConfig, GRPOTrainer
training_args = GRPOConfig(
    vllm_sampling_params = vllm_sampling_params,
    temperature = 1.0,
    learning_rate = 5e-6,
    weight_decay = 0.001,
    warmup_ratio = 0.1,
    lr_scheduler_type = "linear",
    optim = "adamw_8bit",
    logging_steps = 1,
    per_device_train_batch_size = 1,
    gradient_accumulation_steps = 1, # Increase to 4 for smoother training
    num_generations = 8, # Plan B completions are short line-range JSONs — 8 rollouts fit where 4 did before
    max_prompt_length = max_prompt_length,
    max_completion_length = max_completion_length,
    # num_train_epochs = 1, # Set to 1 for a full training run
    max_steps = 100,
    save_steps = 100,
    report_to = "none", # Can use Weights & Biases
    output_dir = "outputs",

    # For optional training + evaluation
    # fp16_full_eval = True,
    # per_device_eval_batch_size = 4,
    # eval_accumulation_steps = 1,
    # eval_strategy = "steps",
    # eval_steps = 1,
)
````

Let's take 20 held-out windows and compare extraction quality with and without our LoRA: the rate of valid line-range JSON and the mean per-section line-overlap F1 against gold (the same metric the content reward uses).

````python
sample_dataset = test_dataset.shuffle(seed = 3407).select(range(20))
sample_dataset
````

````python
def eval_output(output_text, sample):
    """(is_valid_json, mean line-overlap F1 vs gold) for one generated response."""
    pred = _parse_target(_extract_json_text(output_text))
    if pred is None:
        return False, 0.0
    gold = gold_target(sample["answer"])["sections"]
    f1 = _section_line_f1(pred["sections"], gold, sample["line_start"], sample["line_end"])
    return True, f1

stats = {"with": {"valid": 0, "f1": []}, "without": {"valid": 0, "f1": []}}

print("Comparing extraction quality with and without LoRA on 20 windows:")
print("=" * 60)

for i, sample in enumerate(sample_dataset):
    # Each row already carries its own [system, user] extraction prompt.
    text = tokenizer.apply_chat_template(
        sample["prompt"],
        add_generation_prompt = True,
        tokenize = False,
    )

    output_with_lora = model.fast_generate(
        text,
        sampling_params = sampling_params,
        lora_request = model.load_lora("grpo_lora"),
    )[0].outputs[0].text

    output_without_lora = model.fast_generate(
        text,
        sampling_params = sampling_params,
        lora_request = None,
    )[0].outputs[0].text

    for tag, out in (("with", output_with_lora), ("without", output_without_lora)):
        valid, f1 = eval_output(out, sample)
        stats[tag]["valid"] += valid
        stats[tag]["f1"].append(f1)

    # Print progress every 5 samples
    if (i + 1) % 5 == 0:
        print(f"Processed {i + 1}/20 samples...")

print("\n" + "=" * 60)
print("RESULTS:")
for tag, label in (("with", "With LoRA"), ("without", "Without LoRA")):
    n = len(stats[tag]["f1"])
    mean_f1 = sum(stats[tag]["f1"]) / n
    print(f"{label:14s} valid JSON: {stats[tag]['valid']}/{n}   mean line-overlap F1: {mean_f1:.3f}")
````

Our extractor should now emit valid in-window line ranges far more reliably than the base model — it's not always exact, since we only trained for ~100 steps; extend `max_steps` (or switch to `num_train_epochs = 1` over the ~2.5K train windows) for better range F1.
